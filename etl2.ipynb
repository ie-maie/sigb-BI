## Extract

```python
buf = pd.read_csv("buf.csv", encoding='latin1', sep=';')
bua = pd.read_excel("bua.xls", sheet_name="Feuil1")


print("BUF:", buf.shape)
print("BUA:", bua.shape)
```

## Transform

```python
ARABIC_PATTERN = re.compile(r'[\u0600-\u06FF]')

EXPECTED_COLUMNS = [
    "Cote", "Titre", "Auteur", "Lieu", "Edition",
    "Annee", "Nb_pages", "Matiere", "Inventaire"
]

# EXACT ROLES IDENTIFIED IN THE SOURCE BIBLIOGRAPHIES
EXACT_ROLES = [
    # Latin expressions
    r's/s\s*\.?\s*dir\.?', r'S/S\s*dir', r's/sdir', r's\.sdir',
    r'coord', r'cod', r'cor\.', r'Edited\s+by', r'published',
    r'\béd\b\.?', r'\bed\b\.?', r'\bèd\b\.?', r'\bdir\b', r'\bpub\b', r'\bpréf\b', r'\bpré\b', r'\bpref\b',
    # Arabic expressions
    r'إعداد', r'تنسيق', r'إشراف', r'تحت\s+اشراف', r'تحقيق', r'تحرير', r'تحقيق\s+ودراسة'
]

# Regex looking exclusively for the exact roles safely wrapped inside parentheses/brackets
ROLE_REGEX = re.compile(r'[\(\[\{]\s*(?:' + '|'.join(EXACT_ROLES) + r')\s*[\)\]\}]', re.IGNORECASE)

# Structural corporate keywords to prevent split-corruption of institutions
CORPORATE_KEYWORDS = [
    'faculte', 'fac.', 'ministere', 'centre', 'association', 'royaume', 'comite', 'urbama', 'banque',
    'جامعة', 'مركز', 'وزارة', 'جمعية', 'لجنة', 'مؤسسة', 'منظمة', 'إدارة', 'اتحاد', 'إتحاد', 'البنك'
]

# ROBUST MULTI-AUTHOR SPLITTER WITH CONTEXT LOOKBEHINDS
# Splits on explicitly typed separators: ';', '--', ' et ', ' and ', or ' و '
# For period splits (e.g. Name1. Name2), it protects middle initials like "J. K." via negative lookbehind
AUTHOR_SPLIT_REGEX = re.compile(
    r'(?:\s*;\s*|\s*--\s*|\s+et\s+|\s+and\s+|\s+و\s+|(?<=[^\b[A-Z]\b])\.\s+(?=[A-Z\u0600-\u06FF]))'
)


def standardize_columns(df: pd.DataFrame) -> pd.DataFrame:
    """Ensures uniformity of column dimensions matching database fields."""
    df.columns = EXPECTED_COLUMNS
    return df


def clean_cote(df: pd.DataFrame) -> pd.DataFrame:
    """Removes accidental double quotes or whitespace anomalies from Cotes."""
    if "Cote" in df.columns:
        df["Cote"] = df["Cote"].astype(str).str.replace('"', '', regex=False).str.strip()
    return df


def extract_roles_and_split_authors(author_str):
    """
    Analyzes the Auteur entry, extracts the structural role, cleans names
    of remaining symbols, and returns pairs of (Author Name, Extracted Role).
    """
    if pd.isna(author_str) or str(author_str).lower() in ["nan", ""]:
        return []

    author_str = str(author_str).strip()

    # Establish a clean default role fallback context
    detected_role = "Auteur"

    # Isolate explicit metadata roles buried inside brackets
    found_roles = ROLE_REGEX.findall(author_str)
    if found_roles:
        # Extract the pure role token text without brackets
        detected_role = re.sub(r'[\(\[\{\)\]\}]', '', found_roles[0]).strip()
        # Wipe the entire role block snippet from the text string
        author_str = ROLE_REGEX.sub("", author_str)

    # Scan the text string context against institutional names before splitting
    is_corporate = any(kw in author_str.lower() for kw in CORPORATE_KEYWORDS)

    if is_corporate:
        # Strip trailing slashes, commas, or punctuation left behind by metadata boundaries
        auth_cleaned = re.sub(r'^[.,\s\-\/]+|[.,\s\-\/]+$', '', author_str).strip()
        if auth_cleaned:
            return [(auth_cleaned, detected_role)]
        return []

    # Deconstruct multiple composite entities safely
    raw_authors = AUTHOR_SPLIT_REGEX.split(author_str)

    author_role_pairs = []
    for auth in raw_authors:
        # Strict edge stripping to clear symbols left over by role deletion or split bounds
        auth_cleaned = re.sub(r'^[.,\s\(\)\[\]\-]+|[.,\s\(\)\[\]\-]+$', '', auth).strip()

        # Verify block integrity
        if auth_cleaned and auth_cleaned.lower() not in ["nan", "unknown"]:
            author_role_pairs.append((auth_cleaned, detected_role))

    return author_role_pairs


def clean_text_and_extract(df: pd.DataFrame) -> pd.DataFrame:
    """
    Coordinates extraction of roles alongside text sanitization.
    """
    punctuation_sensitive_cols = ["Lieu", "Edition"]

    # Run the paired tuple extractor row-by-row
    pairs_series = df["Auteur"].apply(extract_roles_and_split_authors)

    # Structural column: stores Python native structures: [('Name', 'Role'), ...]
    df["Auteur_Pairs"] = pairs_series

    # Hashable clean string representation for uniqueness checking steps
    df["Auteur_Names"] = pairs_series.apply(lambda pairs: "; ".join([p[0] for p in pairs]))

    # Standard textual cleanups
    for col in ["Titre", "Lieu", "Edition", "Matiere"]:
        df[col] = df[col].astype(str).str.strip()
        if col in punctuation_sensitive_cols:
            df[col] = df[col].str.replace(r'[.,\s]+$', '', regex=True)
        df[col] = df[col].replace(["nan", ""], None)

    return df


def convert_types(df: pd.DataFrame) -> pd.DataFrame:
    """Safely normalizes numeric identifiers to nullable integer classes."""
    for col in ["Annee", "Nb_pages"]:
        df[col] = pd.to_numeric(df[col], errors="coerce")
        df[col] = df[col].astype("Int64")
    return df


def add_language(df: pd.DataFrame, lang_code: str) -> pd.DataFrame:
    """Assigns the explicit ISO-style three-character code matching table constraints."""
    df["language"] = lang_code
    return df


def create_notice_key(df: pd.DataFrame) -> pd.DataFrame:
    """Generates composite keys using unified clean string features."""
    df["notice_key"] = (
        df["Titre"].fillna("") + "|" +
        df["Auteur_Names"].fillna("") + "|" +
        df["Annee"].astype(str).replace('<NA>', '')
    )
    return df


def handle_duplicates(df: pd.DataFrame) -> pd.DataFrame:
    """Tracks and labels duplicate records across key metrics without data loss."""
    # Temporarily drop list structures before calling native duplication checks
    temp_df = df.drop(columns=["Auteur_Pairs"])
    df["is_duplicate_row"] = temp_df.duplicated()
    df["is_duplicate_inventory"] = temp_df.duplicated("Inventaire", keep=False)
    df["is_duplicate_notice"] = temp_df.duplicated("notice_key", keep=False)
    return df


def transform(df: pd.DataFrame, lang_code: str) -> pd.DataFrame:
    """Complete, end-to-end atomic database prep transformation pipeline."""
    df = standardize_columns(df)
    df = clean_cote(df)
    df = clean_text_and_extract(df)
    df = convert_types(df)
    df = add_language(df, lang_code)
    df = create_notice_key(df)
    df = handle_duplicates(df)

    # Realign original Auteur column to store clean string values
    df["Auteur"] = df["Auteur_Names"]
    return df
```

## Load

```python
import ast
import pandas as pd
import numpy as np

def load_to_csv(df: pd.DataFrame, path: str):
    """
    Saves atomic staging files to disk, handling multi-language
    UTF-8 character streams (Arabic/French) without symbol corruption.
    """
    df.to_csv(path, index=False, encoding="utf-8-sig")


def export_for_sql(df: pd.DataFrame):
    """
    Splits the staging DataFrame into clean, normalized structural data tables
    matching your exact relational database schema.

    Handles unique constraints, missing fields, and maps multi-author groups
    to an independent bridge table layout.
    """

    # -------------------------------------------------------------------------
    # 1. NORMALIZED REFERENCE TABLES (LOOKUPS)
    # -------------------------------------------------------------------------
    # Extract unique languages
    langue = df[["language"]].dropna().drop_duplicates()
    langue = langue[langue["language"].astype(str).str.strip() != ""]
    langue.columns = ["code_langue"]

    # Extract unique publishers (Editions)
    editeur = df[["Edition"]].dropna().drop_duplicates()
    editeur = editeur[editeur["Edition"].astype(str).str.strip() != ""]
    editeur.columns = ["nom_editeur"]

    # Extract unique classification codes (Cotes)
    classification = df[["Cote"]].dropna().drop_duplicates()
    classification = classification[classification["Cote"].astype(str).str.strip() != ""]
    classification.columns = ["cote"]

    # Extract unique subjects (Matieres)
    matiere = df[["Matiere"]].dropna().drop_duplicates()
    matiere = matiere[matiere["Matiere"].astype(str).str.strip() != ""]
    matiere.columns = ["libelle"]

    # -------------------------------------------------------------------------
    # 2. SEPARATING AUTHORS & ROLES (MANY-TO-MANY BRIDGE ENGINE)
    # -------------------------------------------------------------------------
    exploded_authors = []
    notice_author_bridge = []

    for _, row in df.iterrows():
        n_key = row.get("notice_key")
        raw_author_pairs_value = row.get("Auteur_Pairs")

        author_pairs_list = []
        if isinstance(raw_author_pairs_value, list):
            author_pairs_list = raw_author_pairs_value
        elif isinstance(raw_author_pairs_value, np.ndarray):
            author_pairs_list = raw_author_pairs_value.tolist()
        elif isinstance(raw_author_pairs_value, str):
            # Attempt to literal_eval if it's a string, e.g., from a saved/loaded DataFrame
            try:
                evaluated = ast.literal_eval(raw_author_pairs_value)
                if isinstance(evaluated, list):
                    author_pairs_list = evaluated
                elif isinstance(evaluated, np.ndarray):
                    author_pairs_list = evaluated.tolist()
            except (ValueError, SyntaxError):
                pass # Not a valid string representation of a list/array

        if not author_pairs_list: # If empty list, or None, or failed conversion
            # Use Auteur_Names as fallback if no structured author pairs could be extracted
            fallback_name = str(row.get("Auteur_Names", "")).strip()
            if fallback_name and fallback_name.lower() not in ["nan", "unknown"]:
                 exploded_authors.append(fallback_name)
                 notice_author_bridge.append({
                     "notice_key": n_key,
                     "nom_complet": fallback_name,
                     "role_auteur": "Auteur" # Default role
                 })
            continue

        # Process the now-guaranteed list of (name, role) tuples
        for name, role in author_pairs_list:
            clean_name = str(name).strip()
            clean_role = str(role).strip() if role else "Auteur"

            if clean_name and clean_name.lower() not in ["nan", "unknown"]:
                exploded_authors.append(clean_name)
                notice_author_bridge.append({
                    "notice_key": n_key,
                    "nom_complet": clean_name,
                    "role_auteur": clean_role
                })

    # Create the single normalized master table of unique authors
    auteur = pd.DataFrame(exploded_authors, columns=["nom_complet"]).drop_duplicates()
    auteur = auteur[auteur["nom_complet"].str.strip() != ""].reset_index(drop=True)

    # Create the relational bridge table linking books to their author entities
    notice_auteur = pd.DataFrame(notice_author_bridge).drop_duplicates().reset_index(drop=True)

    # -------------------------------------------------------------------------
    # 3. CORE NOTICE ENTITY TABLE
    # -------------------------------------------------------------------------
    notice = df[[
        "notice_key",  # Internal schema unique key used for joining indexes
        "Titre",
        "Annee",
        "Nb_pages",
        "Lieu",
        "Edition",
        "language",
        "Cote"
    ]].copy()

    # Drop references that contain missing or null book title items
    notice = notice.dropna(subset=["Titre"])
    notice = notice[notice["Titre"].astype(str).str.strip() != ""]

    # Align column namings perfectly with your SQL Schema definition
    notice.rename(columns={
        "Annee": "annee_pub",
        "Nb_pages": "nb_pages",
        "Lieu": "lieu_edition",
        "Edition": "nom_editeur",       # Used as index text match target during data loading
        "language": "code_langue",      # Directly references your langue setup
        "Cote": "cote"                  # Matches classification lookup string
    }, inplace=True)

    # -------------------------------------------------------------------------
    # 4. PHYSICAL ITEM EXEMPLARY TABLE
    # -------------------------------------------------------------------------
    exemplaire = df[["Inventaire", "notice_key"]].copy()

    # Drop rows missing barcode tags or inventory values to safeguard unique constraints
    exemplaire = exemplaire.dropna(subset=["Inventaire"])
    exemplaire = exemplaire[exemplaire["Inventaire"].astype(str).str.strip() != ""]
    exemplaire.rename(columns={"Inventaire": "num_inventaire"}, inplace=True)

    # Return dictionary payload housing the production-ready DataFrames
    return {
        "langue": langue.reset_index(drop=True),
        "editeur": editeur.reset_index(drop=True),
        "auteur": auteur.reset_index(drop=True),
        "classification": classification.reset_index(drop=True),
        "matiere": matiere.reset_index(drop=True),
        "notice": notice.reset_index(drop=True),
        "notice_auteur": notice_auteur.reset_index(drop=True),
        "exemplaire": exemplaire.reset_index(drop=True)
    }
```
